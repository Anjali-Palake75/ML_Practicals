# 5. Basic Information

```python
titanic['Survived'].value_counts()
```

```python
sns.countplot(data=titanic, x='Survived')
plt.title('Number of Passengers Survived and Died')
plt.xlabel('Survived (0 = No, 1 = Yes)')
plt.ylabel('Number of Passengers')
plt.show()
```

# 6. Passenger Class

```python
titanic['Pclass'].value_counts()
```

```python
sns.countplot(data=titanic, x='Pclass')
plt.title('Passenger Class Distribution')
plt.xlabel('Passenger Class')
plt.ylabel('Number of Passengers')
plt.show()
```

# 7. Gender Distribution

```python
titanic['Sex'].value_counts()
```

```python
sns.countplot(data=titanic, x='Sex')
plt.title('Gender Distribution')
plt.xlabel('Gender')
plt.ylabel('Number of Passengers')
plt.show()
```

# 9. Missing Data

```python
titanic.isnull().sum()
```

# Visualize missing values

```python
plt.figure(figsize=(10, 6))

sns.heatmap(titanic.isnull(),
            yticklabels=False,
            cbar=False)

plt.title('Missing Values in Titanic Dataset')
plt.show()
```

# 10. Data Visualization

# A. Histogram of Age

```python
titanic['Age'].plot(
    kind='hist',
    bins=30
)

plt.title('Distribution of Passenger Age')
plt.xlabel('Age')
plt.ylabel('Frequency')
plt.show()
```

# 11. Box Plot

```python
plt.figure(figsize=(8, 5))

sns.boxplot(data=titanic, x='Age')

plt.title('Box Plot of Passenger Age')
plt.xlabel('Age')
plt.show()
```

# 12. Survival by Gender

```python
sns.countplot(data=titanic, x='Survived', hue='Sex')

plt.title('Survival by Gender')
plt.xlabel('Survived (0 = No, 1 = Yes)')
plt.ylabel('Number of Passengers')
plt.show()
```

# 13. Survival by Passenger Class

```python
sns.countplot(data=titanic, x='Survived', hue='Pclass')

plt.title('Survival by Passenger Class')
plt.xlabel('Survived (0 = No, 1 = Yes)')
plt.ylabel('Number of Passengers')
plt.show()
```

# 14. Scatter Plot – Age vs Fare

```python
sns.scatterplot(
    data=titanic,
    x='Age',
    y='Fare',
    hue='Pclass'
)

plt.title('Age vs Fare by Passenger Class')
plt.xlabel('Age')
plt.ylabel('Fare')
plt.show()
```

# 15. Scatter Plot – Age vs Fare by Gender

```python
sns.scatterplot(
    data=titanic,
    x='Age',
    y='Fare',
    hue='Sex'
)

plt.title('Age vs Fare by Gender')
plt.xlabel('Age')
plt.ylabel('Fare')
plt.show()
```

# 16. Correlation Analysis

```python
correlation = titanic.select_dtypes(include='number').corr()

correlation

correlation['Survived']
```

# 17. Correlation Heatmap

```python
plt.figure(figsize=(10, 7))

sns.heatmap(
    correlation,
    annot=True,
    cmap='crest'
)

plt.title('Correlation Heatmap')
plt.show()
```

# 18. Additional Analysis – Fare Distribution

```python
plt.figure(figsize=(8, 5))

sns.histplot(data=titanic, x='Fare', bins=30)

plt.title('Fare Distribution')
plt.xlabel('Fare')
plt.ylabel('Frequency')
plt.show()
```

# 19. Outlier Detection

```python
plt.figure(figsize=(8, 5))

sns.boxplot(data=titanic, x='Fare')

plt.title('Box Plot of Fare')
plt.xlabel('Fare')
plt.show()
```
