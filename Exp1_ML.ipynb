# Experiment No. 1 Data pre-processing on a raw dataset

# 1. Import libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.preprocessing import MinMaxScaler, StandardScaler
```

# 2. Load Dataset

```python
file_path = r"C:/Users/hp/Downloads/Employee.csv"

df = pd.read_csv(file_path)

print("========== ORIGINAL DATASET ==========")
print(df.head())

print("\nDataset Shape:")
print(df.shape)
```

# 3. Data Information

```python
print("\n========== DATA INFORMATION ==========")
df.info()

print("\n========== STATISTICAL SUMMARY ==========")
print(df.describe())

```

# 4. HANDLING MISSING VALUES

```python
print("\n========== MISSING VALUES BEFORE ==========")
print(df.isnull().sum())

# Numerical columns - Median
df['Age'] = df['Age'].fillna(df['Age'].median())
df['Salary'] = df['Salary'].fillna(df['Salary'].median())

# Categorical columns - Mode
df['Company'] = df['Company'].fillna(df['Company'].mode()[0])
df['Place'] = df['Place'].fillna(df['Place'].mode()[0])
df['Country'] = df['Country'].fillna(df['Country'].mode()[0])

print("\n========== MISSING VALUES AFTER ==========")
print(df.isnull().sum())
```

# 5. HANDLING DUPLICATES

```python
print("\n========== DUPLICATES ==========")

print("Duplicate records before removal:",
      df.duplicated().sum())

df = df.drop_duplicates()

print("Duplicate records after removal:",
      df.duplicated().sum())
```

# 6.OUTLIER DETECTION - AGE

```python
Q1_age = df['Age'].quantile(0.25)
Q3_age = df['Age'].quantile(0.75)

IQR_age = Q3_age - Q1_age

lower_age = Q1_age - 1.5 * IQR_age
upper_age = Q3_age + 1.5 * IQR_age

age_outliers = df[
    (df['Age'] < lower_age) |
    (df['Age'] > upper_age)
]

print("\n========== AGE OUTLIERS ==========")
print("Q1:", Q1_age)
print("Q3:", Q3_age)
print("IQR:", IQR_age)
print("Lower Limit:", lower_age)
print("Upper Limit:", upper_age)
print("Number of Age Outliers:",
      len(age_outliers))
```

# 7. OUTLIER DETECTION - SALARY

```python
Q1_salary = df['Salary'].quantile(0.25)
Q3_salary = df['Salary'].quantile(0.75)

IQR_salary = Q3_salary - Q1_salary

lower_salary = Q1_salary - 1.5 * IQR_salary
upper_salary = Q3_salary + 1.5 * IQR_salary

salary_outliers = df[
    (df['Salary'] < lower_salary) |
    (df['Salary'] > upper_salary)
]

print("\n========== SALARY OUTLIERS ==========")
print("Q1:", Q1_salary)
print("Q3:", Q3_salary)
print("IQR:", IQR_salary)
print("Lower Limit:", lower_salary)
print("Upper Limit:", upper_salary)
print("Number of Salary Outliers:",
      len(salary_outliers))
```

# 8. HANDLE OUTLIERS USING CAPPING

```python
df['Age'] = df['Age'].clip(
    lower=lower_age,
    upper=upper_age
)

df['Salary'] = df['Salary'].clip(
    lower=lower_salary,
    upper=upper_salary
)

print("\nOutliers handled using IQR capping.")
```

# 9. NORMALIZATION

```python
normalizer = MinMaxScaler()

df[['Age_Normalized',
    'Salary_Normalized']] = normalizer.fit_transform(
        df[['Age', 'Salary']]
    )

print("\n========== NORMALIZED DATA ==========")

print(
    df[
        [
            'Age',
            'Salary',
            'Age_Normalized',
            'Salary_Normalized'
        ]
    ].head(10)
)
```

# 10. STANDARDIZATION

```python
standardizer = StandardScaler()

df[['Age_Standardized',
    'Salary_Standardized']] = standardizer.fit_transform(
        df[['Age', 'Salary']]
    )

print("\n========== STANDARDIZED DATA ==========")

print(
    df[
        [
            'Age',
            'Salary',
            'Age_Standardized',
            'Salary_Standardized'
        ]
    ].head(10)
)
```

# 11. VERIFICATION

```python
print("\n========== NORMALIZATION VERIFICATION ==========")

print("Age Min:",
      df['Age_Normalized'].min())

print("Age Max:",
      df['Age_Normalized'].max())

print("Salary Min:",
      df['Salary_Normalized'].min())

print("Salary Max:",
      df['Salary_Normalized'].max())

print("\n========== STANDARDIZATION VERIFICATION ==========")

print("Age Mean:",
      df['Age_Standardized'].mean())

print("Salary Mean:",
      df['Salary_Standardized'].mean())
```

# 11. FINAL DATASET

```python
print("\n========== FINAL PREPROCESSED DATASET ==========")
print(df)
```

# 12. SAVE DATASET

```python
output_file = r"C:\Users\hp\Downloads\Employee_Preprocessed.csv"

df.to_csv(output_file, index=False)

print("\nPreprocessed dataset saved successfully!")
print("Location:", output_file)
```
